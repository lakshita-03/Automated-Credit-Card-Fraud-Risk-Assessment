```python
import pandas as pd
from pathlib import Path

# Locate the project and data folders
project_dir = Path.cwd().parent
data_dir = project_dir / "data"

# Load the datasets
train_df = pd.read_csv(data_dir / "fraudTrain.csv")
test_df = pd.read_csv(data_dir / "fraudTest.csv")

print("Training dataset loaded successfully!")
print("Testing dataset loaded successfully!")

print("\nTraining dataset shape:", train_df.shape)
print("Testing dataset shape:", test_df.shape)
```

### Check data quality

```python
# 1. Check the dataset dimensions
print("Dataset dimensions:", train_df.shape)

# 2. Check for missing values
print("\nMissing values in each column:")
print(train_df.isnull().sum()[train_df.isnull().sum() > 0])

# 3. Check for duplicate rows
print("\nNumber of duplicate rows:", train_df.duplicated().sum())

# 4. Check data types
print("\nData types:")
print(train_df.dtypes)
```

### Understand the fraud distribution

```python
# Count legitimate and fraudulent transactions
fraud_counts = train_df["is_fraud"].value_counts().sort_index()

print("Transaction counts:")
print(fraud_counts)

# Calculate percentages
fraud_percentages = train_df["is_fraud"].value_counts(
    normalize=True
).sort_index() * 100

print("\nTransaction percentages:")
print(fraud_percentages.round(4))
```

### Investigate transaction amounts

```python
print("Transaction amount statistics:")

display(train_df["amt"].describe())

print("\nAverage transaction amount by fraud status:")

display(
    train_df.groupby("is_fraud")["amt"]
    .agg(["count", "mean", "median", "min", "max"])
    .round(2)
)
```

### Transaction count

```python
# Count legitimate and fraudulent transactions
fraud_counts = train_df["is_fraud"].value_counts().sort_index()

print("Transaction counts:")
print(fraud_counts)

# Calculate the percentage of each class
fraud_percentages = (
    train_df["is_fraud"].value_counts(normalize=True).sort_index() * 100
)

print("\nTransaction percentages:")
print(fraud_percentages.round(4))
```

```python
import matplotlib.pyplot as plt

fraud_counts.plot(kind="bar")

plt.title("Distribution of Legitimate and Fraudulent Transactions")
plt.xlabel("Fraud Status (0 = Legitimate, 1 = Fraud)")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### Calculate fraud by merchant category

```python
# Calculate the number of transactions in each category
category_summary = train_df.groupby("category").agg(
    total_transactions=("is_fraud", "size"),
    fraudulent_transactions=("is_fraud", "sum")
)

# Calculate the percentage of transactions labelled as fraud
category_summary["fraud_rate_pct"] = (
    category_summary["fraudulent_transactions"]
    / category_summary["total_transactions"]
) * 100

# Display categories with the highest fraud rates
category_summary = category_summary.sort_values(
    "fraud_rate_pct",
    ascending=False
)

display(category_summary.round(3))
```

### Transaction Amount Analysis
### Legitimate transactions vs Fraudulent transactions

```python
legitimate_mean = train_df.loc[
    train_df["is_fraud"] == 0, "amt"
].mean()

fraud_mean = train_df.loc[
    train_df["is_fraud"] == 1, "amt"
].mean()

print(
    f"Average fraudulent transaction is "
    f"{fraud_mean / legitimate_mean:.2f} times "
    f"the average legitimate transaction."
)
```

```python
import matplotlib.pyplot as plt

# Limit the plot to transactions up to the 99th percentile
# so that extreme values don't compress most of the chart.
amount_limit = train_df["amt"].quantile(0.99)

plot_df = train_df[train_df["amt"] <= amount_limit]

legitimate_amounts = plot_df.loc[
    plot_df["is_fraud"] == 0, "amt"
]

fraud_amounts = plot_df.loc[
    plot_df["is_fraud"] == 1, "amt"
]

plt.figure(figsize=(9, 5))

plt.boxplot(
    [legitimate_amounts, fraud_amounts],
    tick_labels=["Legitimate", "Fraudulent"],
    showfliers=False
)

plt.title("Transaction Amount Distribution by Fraud Status")
plt.ylabel("Transaction Amount")
plt.tight_layout()
plt.show()
```

```python
median_amounts = train_df.groupby("is_fraud")["amt"].median()

median_amounts.index = ["Legitimate", "Fraudulent"]

median_amounts.plot(kind="bar")

plt.title("Median Transaction Amount by Fraud Status")
plt.xlabel("Transaction Type")
plt.ylabel("Median Transaction Amount")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## WHY this difference exists

```python
# Compare transaction amounts across merchant categories
category_amounts = train_df.groupby(
    ["category", "is_fraud"]
)["amt"].median().unstack()

# Rename the columns
category_amounts.columns = [
    "Legitimate_Median",
    "Fraudulent_Median"
]

# Calculate the difference between medians
category_amounts["Median_Difference"] = (
    category_amounts["Fraudulent_Median"]
    - category_amounts["Legitimate_Median"]
)

# Sort by fraudulent median transaction amount
category_amounts = category_amounts.sort_values(
    "Fraudulent_Median",
    ascending=False
)

display(category_amounts.round(2))
```

### Category count analysis

```python
category_counts = train_df.groupby("category").agg(
    total_transactions=("is_fraud", "size"),
    fraudulent_transactions=("is_fraud", "sum")
)

category_counts["fraud_rate_pct"] = (
    category_counts["fraudulent_transactions"]
    / category_counts["total_transactions"]
) * 100

category_counts = category_counts.sort_values(
    "fraud_rate_pct",
    ascending=False
)

display(category_counts.round(3))
```

```python
import matplotlib.pyplot as plt

# Sort categories from highest to lowest fraud rate
plot_data = category_counts.sort_values("fraud_rate_pct")

plt.figure(figsize=(10, 7))

plt.barh(
    plot_data.index,
    plot_data["fraud_rate_pct"]
)

plt.title("Fraud Rate by Merchant Category")
plt.xlabel("Fraud Rate (%)")
plt.ylabel("Merchant Category")
plt.tight_layout()
plt.show()
```

## Analysing the data with transaction time

```python
train_df["trans_date_trans_time"] = pd.to_datetime(
    train_df["trans_date_trans_time"]
)

train_df["transaction_hour"] = (
    train_df["trans_date_trans_time"].dt.hour
)

print(train_df["trans_date_trans_time"].dtype)

display(
    train_df[
        ["trans_date_trans_time", "transaction_hour", "is_fraud"]
    ].head()
)
```

### Calculating the fraud rate by hour

```python
hourly_summary = train_df.groupby("transaction_hour").agg(
    total_transactions=("is_fraud", "size"),
    fraudulent_transactions=("is_fraud", "sum")
)

hourly_summary["fraud_rate_pct"] = (
    hourly_summary["fraudulent_transactions"]
    / hourly_summary["total_transactions"]
) * 100

display(hourly_summary.round(3))
```

```python
plt.figure(figsize=(11, 5))

plt.plot(
    hourly_summary.index,
    hourly_summary["fraud_rate_pct"],
    marker="o"
)

plt.title("Fraud Rate by Transaction Hour")
plt.xlabel("Hour of Day (0–23)")
plt.ylabel("Fraud Rate (%)")
plt.xticks(range(24))
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Extracting days of the week

train_df["day_of_week"] = train_df["trans_date_trans_time"].dt.day_name()

display(
    train_df[["trans_date_trans_time", "day_of_week", "is_fraud"]].head()
)
```

```python
# Calculating fraud rate by day

daily_summary = train_df.groupby("day_of_week").agg(
    total_transactions=("is_fraud", "size"),
    fraudulent_transactions=("is_fraud", "sum")
)

daily_summary["fraud_rate_pct"] = (
    daily_summary["fraudulent_transactions"]
    / daily_summary["total_transactions"]
) * 100

display(daily_summary.round(3))
```

```python
day_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

daily_summary = daily_summary.reindex(day_order)

display(daily_summary.round(3))
```
